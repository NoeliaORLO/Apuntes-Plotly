```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

fig = make_subplots(rows=2, cols=2) # Creamos un subplot con 4 lugares, para luego insertar los gráficos.
```

```python
plot1 = go.Scatter(x=[1,2,3],
                   y=[4,5,6],
                   mode="lines+markers",
                   name="Plot 1")

fig.add_trace(plot1, row=1,col=1)
```

```python
plot2 = go.Bar(x=["A","B", "C"],
               y=[8,9,20],
               name="Plot 2")

fig.add_trace(plot2, row=1, col=2)
```

```python
plot3 = go.Scatter(x=[1,2,3],
                   y=[4,5,6],
                   mode="markers",
                   name="Plot 3")

fig.add_trace(plot3, row=2,col=1)
```

```python
plot4 = go.Bar(x=["X","Y", "Z"],
               y=[10,12,14],
               name="Plot 4")

fig.add_trace(plot4, row=2, col=2)
```

```python
fig.update_layout(title="Subplots",
                  showlegend=False)
```

```python
fig2 = make_subplots(rows=2, cols=2, shared_xaxes=True, shared_yaxes=True) # Ahora tenemos un único eje X y un único eje Y para toda la figura (shared_xaxes y shared_yaxes)

fig2.add_trace(plot1, row=1, col=1)
fig2.add_trace(plot2, row=1, col=2)
fig2.add_trace(plot3, row=2, col=1)
fig2.add_trace(plot4, row=2, col=2)

fig2.update_layout(title="Subplots con ejes compartidos",
                  showlegend=False)
```

```python
# Facetar gráficos
import plotly.express as px

df = px.data.iris() # esta librería también contiene el dataset iris

px.scatter(df,
           x="sepal_width",
           y="sepal_length",
           color="species",
           facet_col="species",# con facet_col se va a crear una columna para cada especie
           title="Gráfico facetado por especie") 
```
