```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(go.Bar(x=[1,2,3,4,5],
                     y=[17,10,14,15,18],
                     name="barras"))

fig.add_trace(go.Scatter(x=[1,2,3,4,5],
                          y=[16,12,11,18,19],
                          mode="lines",
                          name="linea"))

layout = go.Layout(title="Gráfico Personalizado",
                   xaxis_title="Eje X",
                   yaxis_title="Eje Y")

fig.update_layout(layout)
```

```python
# 1. Modificar formato de elementos gráficos del trazado
# Color y ancho de la línea (gráfico scatter)
fig.update_traces(
    selector=dict(name="linea"),
    line=dict(
        color="firebrick", 
        width=4)
)

# Color de la barra (gráfico barras)
fig.update_traces(
    selector=dict(name="barras"),
    marker=dict(color="lightgreen")
)

# Ancho del espacio entre las barras
fig.update_layout(bargap=0.4)
```

```python
# 2. Modificar formato de títulos
# Título principal
fig.update_layout(
    title=dict(
        text="Este es un gráfico<br><sup>Subtipo del gráfico</sup>",
        font=dict(size=24, color="purple", family="Times New Roman"),
        x=0.5,
        xanchor="center" # sitúa el título en el centro del eje de las X
    )
)

# Títulos de Ejes
fig.update_layout(
    xaxis_title=dict(font=dict(size=18, color="blue", family="Courier New")),
    yaxis_title=dict(font=dict(size=18, color="green", family="Courier New"))
)
```

```python
# 3. Modificar formato de los ejes
fig.update_layout(
    xaxis=dict(
        showgrid=True, # muestra las líneas de la cuadrícula del eje X
        gridcolor="lightgray", 
        zeroline=True, # muestra la línea que corresponde al valor 0
        zerolinecolor="gray",
        showline=True, # muestra la línea exterior del eje
        linecolor="black",
        linewidth=2
    ),
    yaxis=dict(
        gridcolor="lightgray",
        zeroline=True,
        zerolinecolor="gray",
        showline=True,
        linecolor="black",
        linewidth=2
    )
)
```

```python
# 4. Modificar parámetros de la leyenda
fig.update_layout(
    legend=dict(
        x=0.3,
        y=0.9,
        bgcolor="rgba(255,255,255,0.5)", # fondo de la leyenda
        bordercolor="black",
        borderwidth=2, # grosor del borde de la leyenda
        font=dict(size=14,color="black", family="Arial"))
    )
```

```python
# 5. Modificar Fondos
fig.update_layout(
    plot_bgcolor="rgba(123,65,789,0.5)", # color de fondo de la zona del gráfico
    paper_bgcolor="lightblue" # color de fondo de toda la figura
)
```

```python
# 6. Modificar Anotaciones
fig.add_annotation(
    x=2,
    y=12,
    text="Punto clave",
    showarrow=True, # muestra una fecha señalando al punto
    arrowhead=5, # tipo de punta de la flecha
    ax=10, # desplazamiento horizontal de la flecha
    ay=-20 # desplazamiento vertical de la flecha
)

fig.add_annotation(
    x=3,
    y=14,
    text="Otro punto clave",
    showarrow=True,
    arrowhead=2,
    ax=-50,
    ay=-50,
    font=dict(
        family="Courier New, monospace",
        size=16,
        color="#ffffff"
    ),
    align="center", # alineación del texto
    arrowcolor="#45e6aa",
    arrowsize=1,
    arrowwidth=2, # grosor de la flecha
    bordercolor="#45e6aa", # color del borde de la anotación
    borderwidth=2, 
    borderpad=4, # espacio entre el texto y el borde
    bgcolor="#666aef", # color de fondo de la anotación
    opacity=0.5
)
```

```python
# 7. Modificar Información Contextual
fig.update_traces(
    selector=dict(name="linea"),
    text=["A", "B", "C", "D", "E"],
    hoverinfo="text" # al pasar el ratón mostramos lo que pone en text
)
```

```python
fig.update_traces(
    selector=dict(name="linea"),
    text=["A", "B", "C", "D", "E"], 
    hovertemplate="X: %{x}<br>Y: %{y}<br>Texto: %{text}<extra>Info</extra>" # definimos exactamente la información que aparece al pasar el ratón
)
```
