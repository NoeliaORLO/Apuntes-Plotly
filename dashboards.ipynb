```python
# Dashboards: tablero que muestra uno o varios gráficos interactivos y permite realizar acciones a través de una aplicación para que el usuario pueda determinar como se va a 
# mostrar la información a través de varios controles.
import dash
from dash import dcc, html

import pandas as pd
import plotly.express as px

# 1. Creamos un DataFrame de ejemplo
df= pd.DataFrame({
    "Fruta":["Manzanas", "Naranjas", "Bananas", "Manzanas", "Naranjas", "Bananas"],
    "Cantidad":[4,1,2,2,4,5],
    "Ciudad": ["BG", "BG", "BG", "BCN", "BCN", "BCN"]
})

# 2. Creamos la aplicación Dash
app = dash.Dash(__name__) # __name__ indica a Dash el módulo actual de Python

# 3. Definimos el layout de la aplicación
app.layout = html.Div(children=[
    html.H1(children="Dashboard Interactivo"), # título del Dashboard

    dcc.Dropdown( # Dropdown crea un menú desplegable que nos permitirá seleccionar una ciudad (label es lo que ve el usuario, y value lo que recibe python)
        id="desplegable",
        options=[
            {"label": "Bogotá", "value": "BG"},
            {"label": "Barcelona", "value": "BCN"}
        ],
        value="BG" # BG será la opción seleccionada por defecto
    ),

    dcc.Graph(id="grafico") # Espacio donde se mostrará el gráfico, que será actualizado por el callback
])

# 4. Creamos un Callback para actualizar el gráfico basado en la selección del dropdown y conectándolo con el graph para hacer los cambios necesarios.
@app.callback(
    dash.Output("grafico", "figure"), # modifica la figura del Graph
    [dash.Input("desplegable", "value")] # recoge el valor seleccionado en el Dropdown
)

# 5. Creamos una función ejecutable cada vez que cambia la ciudad seleccionada
def actualizar_grafico(ciudad_seleccionada):
    df_filtrado = df[df["Ciudad"] == ciudad_seleccionada] # Filtramos el df para quedarnos solo con la ciudad seleccionada
    fig = px.bar(df_filtrado, x="Fruta", y="Cantidad", color="Fruta") # Creamos un gráfico de barras con los datos filtrados
    return fig # Devolvemos el gráfico al Output

# 6. Ejecutamos la aplicación
if __name__ == "__main__":
    app.run(debug=True)
```
